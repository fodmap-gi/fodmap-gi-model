```python
df = df[df['food_time'].isin(['เช้า', 'กลางวัน', 'เย็น'])]

df['date&time'] = pd.to_datetime(df['date&time'], errors='coerce')
df['date&time'] = df['date&time'].dt.normalize()

period_order = {
    'เช้า': 0,
    'กลางวัน': 1,
    'เย็น': 2
}

df['period_order'] = df['food_time'].map(period_order)

df = df.sort_values(
    ['user_id', 'date&time', 'period_order']
).reset_index(drop=True)

df['Actual_fodmap'] = 0

same_day = (
    (df['date&time'] == df.groupby('user_id')['date&time'].shift(-1)) &
    (df['period_order'] + 1 == df.groupby('user_id')['period_order'].shift(-1))
)

next_day = (
    (df['date&time'] + pd.Timedelta(days=1)
     == df.groupby('user_id')['date&time'].shift(-1)) &
    (df['food_time'] == 'เย็น') &
    (df.groupby('user_id')['food_time'].shift(-1) == 'เช้า')
)

fodmap_check = df.groupby('user_id')['ynbloating'].shift(-1) == 1

df.loc[(same_day | next_day) & fodmap_check, 'Actual_fodmap'] = 1
```
